```python
import pandas as pd
df = pd.read_csv('custdata.tsv', sep='\t')

df.head(10)
```

```python
df.isna().sum()
```

```python
df=df.dropna(subset=["housing.type"])
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

counts= df['housing.type'].value_counts()

counts.plot(kind="bar", color="orange", edgecolor="black")
plt.title("Number of Customers Per Housing Type")
plt.xlabel("Housing Type")
plt.ylabel("Number of Customers")
plt.xticks(rotation=30, ha="right")
plt.show()
```

56 rows containting no data from the housing type were removed to make this bar chart.
