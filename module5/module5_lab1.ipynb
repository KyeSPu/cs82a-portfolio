```python
import pandas as pd
chick_df= pd.read_csv('chickwts.csv')
chick_df.head(10)
```

```python
chick_df.groupby("feed")["weight"].mean().sort_values()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.barplot(data=chick_df, x="feed", y="weight")
plt.title("Average weight of all feed types")
plt.xlabel("Feed")
plt.ylabel("Weight in grams (g)")
plt.show()
```

```python
sns.histplot(data=chick_df, x="weight", bins=15)
plt.title("Frequency of Chicks according to Weight")
plt.xlabel("Weight in grams (g)")
plt.ylabel("Number of Chicks")
plt.show()
```

```python
visitor_df= pd.read_csv('museum_visitors.csv')
visitor_df['Date'] = pd.to_datetime(visitor_df['Date'])
sns.lineplot(data=visitor_df, x="Date", y="Chinese American Museum")
plt.title("Chinese American Museum Visitors Over Time")
plt.xlabel("Date")
plt.ylabel("Visitors")
plt.show()
```

Visitors to the Chinese American Museum generally peak in the first half of the year every year.The number of visitors increased from 2014 until it peaked with the highest record being in 2016 where it steadily decreases from 2017 to 2019 but still the highest amount of visitors still remains higher than those of 2014 to 2015.

The null hypothesis between the weight of the sunflower-fed and horsebean-fed chicken suggests that there is no difference between the two.

```python
from scipy.stats import ttest_ind
a = chick_df[chick_df["feed"] == "sunflower"]["weight"]
b = chick_df[chick_df["feed"] == "horsebean"]["weight"]
ttest_ind(a, b)
```

The difference in grams of the mean is 328.92-160.2 which is roughly 168.72 as 328.92 is rounded to 2 decimal places. The p-value is 2.37 e^-8 which is less than 0.5 suggesting that the null hypothesis should be rejected. This means that there is a significant difference between feeding a chicken sunflower and horsebean seeds. Some things to note are that this is just a small sample size, meaning there could be some deviation that hasn't been recorded, and that this is one farm's data, which could mean it might not be reproducable elsewhere until tests on other farms confirm there is a significant difference

For the bar graph, it started at zero, for the y axis to make sure the data is honest. Not only that, the interval increases by 50 grams every time, instead of making inconsistent numbers to purposely distort the real value that the data has.
