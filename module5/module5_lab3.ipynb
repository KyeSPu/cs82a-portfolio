Getting rid of bad data:

Remove rows with ages equal to 0 or over 100 because those ages are unreasonable, remove negative values for income since income is inherently a positive value, for number of vehicles, remove NA's since those should not be included in the overall calculations

```python
df_clean = df[(df["age"]>0) & (df["age"] <= 100)]
df_clean = df_clean[df_clean["income"] >= 0]
df_clean = df_clean.dropna(subset=["num.vehicles"])
```
