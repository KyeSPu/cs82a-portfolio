```python
import pandas as pd
df= pd.read_csv("titanic_train.csv")
df.isna().sum()
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
df.isna().sum()
```

Median is a good choice as a replacement for NA since it is more resistant to extreme values. 177 values replaced with the median

```python
df["Sex"] = df["Sex"].map({"male": 0, "female": 1})

X = df[["Pclass", "Sex", "Age", "Fare"]]
y = df["Survived"]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
print(X_train.shape)
print(X_test.shape)
```

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(max_depth=4, random_state=42)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

test_accuracy = model.score(X_test, y_test)
print(test_accuracy)
```

```python
results = X_test.copy()
results["Actual"] = y_test
results["Predicted"] = y_pred

errors = results[results["Actual"] != results["Predicted"]]

print("\nNumber of incorrect predictions:", len(errors))
print("\nMisclassified passengers:")
print(errors)
```

The decision tree result I got achieved a test accuracy of 82.46%, which means that it can reasonably predict Titanic passenger survival based on passenger class, sex, age, and fare. The males are the ones which are misclassified the most as most of the misclassification determined by sex is 0, which is male.
