```python
import pandas as pd
df= pd.read_csv("ratings.tsv", sep="\t")

df.head()
```

```python
df.shape
```

```python
df.columns
```

```python
from sklearn.linear_model import LinearRegression

X = df[["ambience", "food", "service"]]
y = df["rating"]

model= LinearRegression()
model.fit(X, y)

for name, coef in zip(X.columns, model.coef_):
    print(name, ":", round(coef, 4))
```

```python
coefficients = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": model.coef_
})

coefficients = coefficients.sort_values(
    by="Coefficient", ascending=False
)

print(coefficients.to_string(index=False))
```

The higher the coefficient the more likely it impacts the rating of the restaurant. In this case, Food is a clear recommendation for what to focus on when wanting a higher rating in a struggling restaurant with a coefficient of about 0.73 on food. Service is next with around 0.32 and ambience is last at 0.13 which are massively small values compared to the food.
