```python
import pandas as pd
df=pd.read_csv("sales_clean.csv")

df.shape
```

```python
print(df["price"].mean())
print(df["price"].median())
print(df["price"].std())

print(df["qty"].mean())
print(df["qty"].median())
print(df["qty"].std())
```

I would take the price median over the mean because the standard deviation is so high at about 72.1 . A high standard deviation implies that there are extreme outliers which affects the mean massively making the mean an unreliable source of data.

```python
df["price"].plot(kind="hist", bins=20)
```

The data for price is skewed right on the histogram which means most of the price of sales were below $100.

```python
df["revenue"]=df["price"]*df["qty"]
df.head()
```

```python
df.corr(numeric_only=True)
```

The data suggests that revenue and price have the higher correlation with a .82 correlation.

```python
df.plot.scatter(x="price", y="revenue")
```

The data from the scatter plot seems to suggest that the higher the price, the more associated it is with revenue. However, from our histogram, we learn that sales are more frequent when price is lower than 100. In fact, the scatter plot supports this outlook with most of the plots located under the price of 100 and being sparser as price increases. There is a obvious correlation betwen revenue and price, the price directly affects the revenue as revenue is determined from price multiplied by quantity. However, someone would incorrectly assume that the greater the price, the greater the revenue which is true, but the revenue is only for individual sales, nothing about the frequency or the total revenue obtained. The little sales when prices are low are much more frequent.
