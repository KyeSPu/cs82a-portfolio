Mean weight of the books by hand:

(800+950+1050+350+750+600+1075+250+700+650+975+350+950+425+725)/15 = 10600/16= 706.67 rounded to 2 decimal places

```python
print(df["weight"].mean())
```

This shows the mean of all the books combined is the same as if done by hand, approximately 706.67 grams

```python
df.groupby("cover")["weight"].mean()
```

The average weight of the hard cover books is approximately 796.43 grams.
The average weight of paperback books is 628.125 grams.

This shows that the average hard cover book weighs more than an average paperback book. This means that for the library scenario proposed in the problem, they might want to consider switching to all paperbacks as they will weigh significantly less than those of a hardcover copy.
