```python
import pandas as pd
df=pd.read_csv("messy_sales.csv")

df.shape
```

```python
df.head(10)
```

```python
df.info()
```

One error is that price only has 288 non-null values with some of them being NaN, that is 12 values unaccounted for. The dates are not in order and have an inconsistent format. Zip code seems to have an error, with 2134 being a 4 digit instead of a typical 5 digit number.

```python
df.isna().sum()
```

```python
fill_value = df["price"].median()
df["price"]=df["price"].fillna(fill_value)

df.isna().sum()
```

```python
print(df.duplicated().sum())
df = df.drop_duplicates()

df.shape
```

```python
df["zip"] = df["zip"].astype(str).str.zfill(5)
df.head(10)
```

```python
df["date"] = pd.to_datetime(df["date"], format="mixed")
df.dtypes
```

```python
df[df["qty"] <= 0]
```

```python
df=df[df["qty"] > 0]
```

```python
df.to_csv('sales_clean.csv', index=False)
```

Cleaning Log

12 columns of prices filled - This is because no value was previously assigned, the median was added to make the data more accurate compared to the mean.

8 Duplicate Results found and dropped - The repeating data are removed because the transaction was likely recorded twice accidentally

Zips become 5 digits - The original file didn't use the correct format for ZIP code so the 0 was added to the beginning to make it 5 digit.

Dates went from str to datetime - The date was treated as a string instead of as a date, the mixed allows the dates to function as is

Negative qtys dropped - The data here is sales data, zero and negative values for quantities are not sales data and should be removed from this list and recorded elsewhere.
