```python
def multiply(first, second):
    total=0
    while second>0:
        total+=first
        second-=1
    return total
print(multiply(7,6))
```

The while loop will continue to add while the second number decreases by 1 eventually reaching to 0 in which the loop will terminate returning the total of first number added by the number of loops (which is the second number).

```python
def multiply(first, second):
    total=0
    for i in range(second):
        total+=first
    return total
print(multiply(7,6))
```
