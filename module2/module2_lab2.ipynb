```python
def multiply(first, second):
    total=0
    counter=0
    
    while counter<abs(second):
        total+=first
        counter+=1
        
    if second<0 :
        return -total
    
    return total
    
print(multiply(7,-6))
```

My original code wouldn't have worked for negative numbers, instead this code works because taking the absolute value would allow the number of additions to be the same for both positive and negative values, this is why the counter variable is needed, to add a value of 1 each time until is becomes the value of abs(second) where the while loop terminates. The if statement at the end determines if the result should be the negative since the absolute value removes the negative sign for the loop.

```python
def multiply(first, second):
    total=0
    for i in range(abs(second)):
        total+=first
        
    if second<0 :
        return -total
    
    return total
    
print(multiply(7,-6))
```
