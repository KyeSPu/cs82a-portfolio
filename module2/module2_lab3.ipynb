```python
#Create Lists
temperatures=[10, 16, 23, 28, 14, 22, 30]
cold=[]
warm=[]
hot=[]

#Create a loop that goes through each value in temperature and assigns them to their respective temperatures
for i in temperatures:

    #When the temperature is less than 15, add to cold
    if i<15:
        cold.append(i)

    #When the temperature is between 15 and 25 (including at 15 and 25), add to warm
    elif i>=15 and i<=25:
        warm.append(i)

    #When the temperature is over 25, add to hot
    elif i>25:
        hot.append(i)

#print out all the values 
print(f"The cold values are {cold}")
print(f"The warm values are {warm}")
print(f"the hot values are {hot}")
```
